**Ideal Activation Function**



1.   Activation Function should be non-linear
2.   Activation Function should be differentiable (Gradient discent should be applicable)
3.   Activation Function should be computationally inexpensive
4.   Activation Function should be zero centred
5.   Activation Function should be non-saturating (otherwise gradiant discent problem appears)



**Sigmoid Function**: The sigmoid function is a smooth, S-shaped curve that maps any real-valued number to a value between 0 and 1. It is often used in the output layer of a binary classification model, where the goal is to predict a binary outcome (e.g., true or false, 0 or 1).

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
def Sigmoid(x):
  return 1 / (1 + np.exp(-x))

x = np.linspace(-10, 10, 100)
y = Sigmoid(x)

plt.plot(x, y)
plt.xlabel('x')
plt.ylabel('Sigmoid(x)')
plt.title('Sigmoid Function')
plt.grid()
plt.show()
```

Advantages:
1.   Used before binary classification
2.   Non-linear activation function
3.   Differentiable

Disadvantages:
1.   Saturated (results in vanishing gradient problem)
2.   Not zero-centred
3.   Computationally Expensive




**Tanh**: The hyperbolic tangent function is similar to the sigmoid function, but it maps values to a range between -1 and 1. It is often used in the hidden layers of neural networks.

```python
def Tanh(x):
  return np.tanh(x)

x = np.linspace(-10, 10, 100)
y = Tanh(x)

plt.plot(x, y)
plt.xlabel('x')
plt.ylabel('Tanh(x)')
plt.title('Tanh Function')
plt.grid()
plt.show()
```

Advantages:
1.   Non-linear activation function
2.   Differentiable
3.   Zero Centred

Disadvantages:
1.   Saturated (results in vanishing gradient problem)
2.   Computationally expensive



**ReLU**:

```python
def ReLU(x):
  return np.maximum(0, x)

x = np.linspace(-10, 10, 100)
y = ReLU(x)

plt.plot(x, y)
plt.xlabel('x')
plt.ylabel('ReLU(x)')
plt.title('ReLU Function')
plt.grid()
plt.show()
```

Advantages:
1.   Non-saturated
2.   Computationally inexpensive
3.   Differentiable
  *  if x < 0; 0
  *  if x >= 0; 1
4.   Non-linear

Disadvantages:
1.   Not zero centred
2.   Dying ReLU Problem
  *   dead neuron: a neuron is said to be dead for any input, it always gives the output 0
Reasons for Dying ReLU Problem

  *   high-learning rate
  *   high-negative bias





**Variants of ReLU**:



1.  **Linear ReLU**
  *   Leaky ReLU
  *   Parametric ReLU


2.   **Non-linear ReLU**
  *   ELU (Exponential Linear Unit)
  *   SeLU (Scaled Exponential Linear Unit)





```python
def LeakyReLU(x, alpha = 0.01):
  return np.where(x > 0, x, alpha * x)

x = np.linspace(-10, 10, 100)
y = LeakyReLU(x)

plt.plot(x, y)
plt.xlabel('x')
plt.ylabel('ReLU(x)')
plt.title('Leaky ReLU Function')
plt.grid()
plt.show()
```

Advantages:
1.   Non-saturated
2.   Computationally inexpensive
3.   Non - linear
4.   No dying ReLU problem
5.   Close to zero centred





**Parametric ReLU:**

```python
def parametricReLU(x, alpha = 0.1):
  return np.where(x > 0, x, alpha * x)

x = np.linspace(-10, 10, 100)
y = parametricReLU(x)

plt.plot(x, y)
plt.xlabel('x')
plt.ylabel('ReLU(x)')
plt.title('Parametric ReLU Function')
plt.grid()
plt.show()
```

Advantages:
1.   Non-saturated
2.   Computationally inexpensive
3.   Non - linear
4.   No dying ReLU problem
5.   Close to zero centred

**ELU:**

```python
def ELU(x, alpha = 0.1):
  return np.where(x > 0, x, alpha * (np.exp(x) - 1))

x = np.linspace(-10, 10, 100)
y = ELU(x)

plt.plot(x, y)
plt.xlabel('x')
plt.ylabel('ELU(x)')
plt.title('ELU Function')
plt.grid()
plt.show()
```

Advantages:
1.   Non-saturated
2.   Non - linear
3.   No dying ReLU problem
4.   Close to zero centred

Disadvantages:
1. Computationally expensive

**SELU:**

```python
def SELU(x, lmbda = 1.0507, alpha = 1.67326):
  return lmbda * np.where(x > 0, x, alpha * (np.exp(x) - 1))

x = np.linspace(-10, 10, 100)
y = SELU(x)

plt.plot(x, y)
plt.xlabel('x')
plt.ylabel('SELU(x)')
plt.title('SELU Function')
plt.grid()
plt.show()
```

Advantages:
1.   Non-saturated
2.   Computationally inexpensive
3.   Non - linear
4.   No dying ReLU problem
5.   Close to zero centred
6.   Convergance is faster
7.   self-normalization

**SoftMax**: The softmax function is used in the output layer of a multi-class classification model, where the goal is to predict one of several possible classes. It takes a vector of real-valued numbers as input and normalizes it into a probability distribution over the classes.

```python
def SoftMax(x):
  exp_x = np.exp(x)
  return exp_x / np.sum(exp_x)


x = np.linspace(-10, 10, 100)
y = SoftMax(x)

plt.plot(x, y)
plt.xlabel('x')
plt.ylabel('SoftMax(x)')
plt.title('SoftMax Function')
plt.grid()
plt.show()
```

Advantages:
1.   Non-saturated
2.   Non - linear
3.   Close to zero centred

Disadvantages:
1. Computationally expensive

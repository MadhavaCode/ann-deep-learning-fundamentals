```python
# Import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
import seaborn as sns
from mlxtend.plotting import plot_decision_regions

import tensorflow
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.layers import Dropout
from tensorflow.keras.optimizers import Adam

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler

import tensorflow
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense, Flatten

from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv('/content/admission_data.csv')
```

```python
x = df.drop('Chance of Admit ', axis=1)
y = df.iloc[:, -1]
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
scaler = MinMaxScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
model = Sequential()
model.add(Dense(10, activation='relu', input_dim=7))
model.add(Dense(10, activation='relu'))
model.add(Dense(1, activation='linear'))
```

```python
model.summary()
```

```python
model.compile(loss='mean_squared_error', optimizer='Adam')
```

```python
history = model.fit(x_train, y_train, epochs=50, validation_split=0.2)
```

```python
plt.plot(history.history['loss'], label = "Loss")
plt.plot(history.history['val_loss'], label = "Validation Loss")

plt.legend()
plt.title('Model Loss')
plt.ylabel('Accuracy')
plt.xlabel('Epoch')
```

```python
model1 = Sequential()
model1.add(Dense(10, activation='relu', input_dim=7, kernel_regularizer = tensorflow.keras.regularizers.l1(0.01)))
model1.add(Dense(10, activation='relu', kernel_regularizer = tensorflow.keras.regularizers.l1(0.01)))
model1.add(Dense(1, activation='sigmoid'))
```

```python
model1.summary()
```

```python
model1.compile(loss='mean_squared_error', optimizer='Adam')
```

```python
history1 = model1.fit(x_train, y_train, epochs=50, validation_split=0.2)
```

```python
plt.plot(history1.history['loss'], label = "Loss")
plt.plot(history1.history['val_loss'], label = "Validation Loss")

plt.legend()
plt.title('Model Loss')
plt.ylabel('Accuracy')
plt.xlabel('Epoch')
```

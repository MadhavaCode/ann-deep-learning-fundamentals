```python
# Import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
import seaborn as sns
from mlxtend.plotting import plot_decision_regions
import tensorflow
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.layers import Dropout
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.regularizers import l2
```

```python
x, y = make_moons(n_samples=500, noise=0.25, random_state=0)
```

```python
plt.scatter(x[:,0], x[:,1], c=y)
plt.show()
```

```python
model = Sequential()
model.add(Dense(128, input_dim=2, activation='relu'))
model.add(Dense(128, activation='relu'))
model.add(Dense(1, activation='sigmoid'))
```

```python
model.summary()
```

```python
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
history = model.fit(x, y, epochs=1000, validation_split = 0.2, verbose=0)
```

```python
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()
plt.show()
```

```python
model1 = Sequential()
model1.add(Dense(128, input_dim=2, activation='relu', kernel_regularizer = tensorflow.keras.regularizers.l1(0.01)))
model1.add(Dense(128, activation='relu', kernel_regularizer = tensorflow.keras.regularizers.l1(0.01)))
model1.add(Dense(1, activation='sigmoid'))
```

```python
model1.summary()
```

```python
model1.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
history1 = model1.fit(x, y, epochs=1000, validation_split = 0.2, verbose=0)
```

```python
plt.plot(history1.history['loss'], label='Training Loss')
plt.plot(history1.history['val_loss'], label='Validation Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()
plt.show()
```

```python
model2 = Sequential()
model2.add(Dense(128, input_dim=2, activation='relu', kernel_regularizer = tensorflow.keras.regularizers.l2(0.01)))
model2.add(Dense(128, activation='relu', kernel_regularizer = tensorflow.keras.regularizers.l2(0.01)))
model2.add(Dense(1, activation='sigmoid'))
```

```python
model2.summary()
```

```python
model2.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
history2 = model2.fit(x, y, epochs=1000, validation_split = 0.2, verbose=0)
```

```python
plt.plot(history2.history['loss'], label='Training Loss')
plt.plot(history2.history['val_loss'], label='Validation Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()
plt.show()
```

```python
plot_decision_regions(x, y, clf=model)
plt.xlim(-2, 3)
plt.ylim(-1.5, 2)
plt.show()
```

```python
plot_decision_regions(x, y, clf=model1)
plt.xlim(-2, 3)
plt.ylim(-1.5, 2)
plt.show()
```

```python
plot_decision_regions(x, y, clf=model2)
plt.xlim(-2, 3)
plt.ylim(-1.5, 2)
plt.show()
```
